# 1. Attribute completeness

### Requirements

1. **Settings:** the required attributes list per class.
2. **Check each rule per feature:** one column per rule where True or False values are stored if the rule applies. For example has_name. Values like surface or religion are checked in the tags column, which holds several feature's tags.
3. **Feature score:** the share of rules it passes, from 0 to 1. A road with a name but no surface gets 0.5.
4. **Class score:** the average of its feature scores. This is the S for attribute completeness in the DQI in equation (1).
5. **Save the result:** a completeness table (feature_id, rule columns, score) and the errors (feature_id plus the rules it failed), so there's easy access or way to filter the failing features in QGIS.

### 1.1 Defining the attribute list and the rules

```python
MAIN_ROADS = ["motorway", "trunk", "primary", "secondary"]   # updated according the meeting 10/11/2026

PLACEHOLDERS = ["road", "fixme", "fix", "name", "updateme"]   # values that count as missing

RULES = {
    "roads": [
        {"rule": "has_name_or_ref", "tags": ["name", "ref"], "applies_to": {"highway": MAIN_ROADS}, "missing_values": PLACEHOLDERS},
        {"rule": "has_surface", "tags": ["surface"], "applies_to": {"highway": MAIN_ROADS}},
    ],
    "buildings": [
        {"rule": "has_building_type", "tags": ["building"], "applies_to": None, "missing_values": ["yes"]},
    ],
    "pois": [
        {"rule": "has_name", "tags": ["name"], "applies_to": None, "missing_values": PLACEHOLDERS},
        {"rule": "has_religion", "tags": ["religion"], "applies_to": {"amenity": ["place_of_worship"]}},
    ],
}
```

```python
def check_rule(tags, rule):

    # 1. check if the rule apply to this feature
    conditions = rule["applies_to"]
    if conditions is not None:
        for key, allowed_values in conditions.items():
            feature_value = tags.get(key)
            if feature_value not in allowed_values:
                return None   # rule does not apply

    # 2. values that count as missing (if there are)
    if "missing_values" in rule:
        missing_values = rule["missing_values"]
    else:
        missing_values = []

    # 3. the rule passes if any of its tags has a useful value
    for tag_name in rule["tags"]:
        value = tags.get(tag_name)
        if value is None:
            continue   # tag not present, try the next one

        clean_value = value.strip().lower()   
        if clean_value == "":
            continue   # empty value (or only spaces), try the next one

        if clean_value in missing_values:
            continue   # placeholder, try the next one

        return True    # useful value found

    return False       # no tag had a useful value
```

Check the rules to all of the roads

```python
roads = gpd.read_file(GPKG_PATH, layer="roads")
tags = roads["tags"].apply(json.loads)

rule_cols = []

for rule in RULES["roads"]:
    results = []                          # one result per road
    for one_road_tags in tags:            # tags = the tags of every road
        result = check_rule(one_road_tags, rule)
        results.append(result)
    column_name = rule["rule"]            # e.g. "has_name_or_ref"
    roads[column_name] = pd.array(results, dtype="boolean")
    rule_cols.append(column_name)

# structure the passed and applicable rules and calculate the score per road
roads["passed"] = roads[rule_cols].sum(axis=1)
roads["applicable"] = roads[rule_cols].notna().sum(axis=1)
roads["score"] = roads["passed"] / roads["applicable"]
roads.head(3)
```

Show the summary of the rules assessed

```python
for column_name in rule_cols:
    applied = roads[column_name].notna().sum()
    passed = roads[column_name].sum()
    if applied == 0:
        print(f"{column_name}: rule did not apply to any feature")
    else:
        share = passed / applied
        print(f"{column_name}: {passed} of {applied} pass ({share:.2%})")

scored_roads = roads[roads["applicable"] > 0]
print(f"Roads scored: {len(scored_roads)}")
```

Calculating roads score by length

```python
# 1. To get the result according the length (m, not degrees), suggested in the meeting 10/11/2026
roads_metric = roads.to_crs(roads.estimate_utm_crs())
roads["length_m"] = roads_metric.length

# 2. keep only roads where at least one rule applies
scored_roads = roads[roads["applicable"] > 0]

# 3. weighted score = sum(score x length) / sum(length)
weighted_sum = 0
total_length = 0
for index, road in scored_roads.iterrows():
    weighted_sum = weighted_sum + road["score"] * road["length_m"]
    total_length = total_length + road["length_m"]

class_score_roads = weighted_sum / total_length
print("Class score (roads, by count):", roads["score"].mean())
print("Class score (roads, by length):", class_score_roads)
```

### 2.1 Growth and event detection

Set parameters

```python
# --- settings: here the parameters are gonna be stablished together with UN ---
Z_THRESHOLD = 2.33     # 99% one-sided: a residual this many standard deviations above average is an event
IMPORT_SHARE = 0.84    # one user made more than 84% of the edits -> import
SIZE_SHARE = 0.10      # one month adds more than 10% of today's total -> event (extra rule, see below)

# which series column and which history flag belong to each class 
CLASS_COLUMNS = {
    "roads": ("road_km", "is_road"),
    "buildings": ("buildings", "is_building"),
    "pois": ("pois", "is_poi"),
}
```

The curve:
a = start level
b = growth size
c = speed
d = month of fastest growth

```python
def logistic(t, a, b, c, d):
    return a + b / (1 + np.exp(-c * (t - d)))
```

```python
month_number = np.arange(len(growth))    
fit_quality_rows = []      # one row per class: how well the logistic curve fits
event_rows = []

for class_name, (series_column, class_flag) in CLASS_COLUMNS.items():
    values = growth[series_column].to_numpy(dtype=float)
    
    # 1. fit the logistic curve
    first_guess = [values.min(), values.max() - values.min(), 0.1, len(values) / 2]
    try:
        parameters, covariance = curve_fit(logistic, month_number, values, p0=first_guess, maxfev=20000, method="lm")  
    except RuntimeError:
        print(f"{class_name}: the curve could not be fitted")
        continue
    fitted = logistic(month_number, *parameters)

    # 2. lagged residuals: how much did each month grow more than the curve expected
    residuals = values - fitted
    lagged_residuals = np.diff(residuals, prepend=residuals[0]) 
    mean_lagged_residuals = lagged_residuals.mean()
    std_lagged_residuals = lagged_residuals.std()
    z_scores = (lagged_residuals - mean_lagged_residuals) / std_lagged_residuals

    ## plot the residuals
    # plt.hist(lagged_residuals, bins=30, edgecolor="black")
    # plt.xlabel("Lagged Residuals")
    # plt.ylabel("Frequency")
    # plt.title(f"Residuals of {class_name}")
    # plt.show()

    # fit quality (Grinberger et al. 2021): NRMSE = RMSE as % of the median observed value
    squared_residuals = residuals ** 2
    rmse_events_fit = np.sqrt(squared_residuals.mean())
    median_value = np.median(values)                       # middle value of the series
    if median_value > 0:
        nrmse_events_fit = rmse_events_fit / median_value * 100
    else:
        nrmse_events_fit = None                            # almost empty series

    fit_quality_rows.append({
        "class": class_name,
        "rmse": rmse_events_fit,
        "nrmse_percent": nrmse_events_fit,
    })

    
    # 3. check every month with both rules
    final_value = values[-1]
    for i in range(1, len(values)):
        increase = values[i] - values[i - 1]
        share_of_final = increase / final_value
        rule_residual = z_scores[i] > Z_THRESHOLD
        rule_size = share_of_final > SIZE_SHARE
        if not rule_residual and not rule_size:
            continue

        # 4. who made the edits of this class in that month?
        month_start = growth["month"].iloc[i]
        cutoff = month_start + pd.DateOffset(months=1)
        in_month = history[(history["timestamp"] >= month_start)
                           & (history["timestamp"] < cutoff)
                           & (history[class_flag])]
        if len(in_month) == 0:
            top_user = None
            top_user_share = None
            event_type = "unknown (no edits found)"
        else:
            user_counts = in_month["user"].value_counts()
            top_user = user_counts.index[0]
            top_user_share = user_counts.iloc[0] / len(in_month)
            if top_user_share > IMPORT_SHARE:
                event_type = "import (one user)"
            else:
                event_type = "campaign or other (many users)"

        event_rows.append({
            "class": class_name, "month": month_start, "increase": increase,
            "share_of_final": share_of_final, "z_score": z_scores[i],
            "rule_residual": rule_residual, "rule_size": rule_size,
            "top_user": top_user, "top_user_share": top_user_share,
            "event_type": event_type,
        })


```

Settings: here the parameters are gonna be stablished together with UN 

```python
RECENT_MONTHS = 12          # window to measure recent growth
RECENT_GROWTH_MAX = 0.03    # saturated if organic growth in the last 12 months < 3%
STABILITY_MAX = 0.10        # curves agree if their totals differ by less than 10%
RMSE_TOLERANCE = 1.5        # "fits equally well" = error at most 1.5 x the best curve's error
```

Define the functions to fit the growth

```python
# --- the candidate curves, where each of them returns the value at month t ---
def logistic(t, a, b, c, d):
    return a + b / (1 + np.exp(-c * (t - d)))

def gompertz(t, a, b, c, d):
    return a + b * np.exp(-np.exp(-c * (t - d)))

def asymptotic(t, a, b, c):
    return a - b * np.exp(-c * t)
```

Organic series: remove each event's increase from its month onward

```python
month_number = np.arange(len(growth))
last_month = month_number[-1]


organic = growth.copy()
event_volume = {}          # total removed per class, added back at the end
for class_name, (series_column, class_flag) in CLASS_COLUMNS.items():
    total_removed = 0
    class_events = events[events["class"] == class_name]
    for index, event in class_events.iterrows():
        from_event_on = organic["month"] >= event["month"]
        organic.loc[from_event_on, series_column] = organic.loc[from_event_on, series_column] - event["increase"]
        total_removed = total_removed + event["increase"]
    event_volume[class_name] = total_removed
```

Fit every curve to every class

```python
fit_rows = []
for class_name, (series_column, class_flag) in CLASS_COLUMNS.items():
    values = organic[series_column].to_numpy(dtype=float)
    current_organic = values[-1]

    # recent growth, measured on the data (the same for every curve)
    value_a_year_ago = values[-1 - RECENT_MONTHS]
    if current_organic > 0:
        recent_growth = (current_organic - value_a_year_ago) / current_organic
    else:
        recent_growth = None

    candidates = [
        ("logistic", logistic, [values.min(), values.max() - values.min(), 0.1, len(values) / 2]),
        ("gompertz", gompertz, [values.min(), values.max() - values.min(), 0.1, len(values) / 2]),
        ("asymptotic", asymptotic, [values.max(), values.max() - values.min(), 0.01]),
    ]
    for curve_name, curve, first_guess in candidates:
        try:
            parameters, covariance = curve_fit(curve, month_number, values,
                                               p0=first_guess, maxfev=20000)  
        except RuntimeError:
            fit_rows.append({"class": class_name, "curve": curve_name, "fitted": False})
            continue

        fitted = curve(month_number, *parameters)
        rmse = np.sqrt(np.mean((values - fitted) ** 2))   # typical error of the fit

        # NRMSE, but on the organic series (events removed)
        median_organic = np.median(values)
        if median_organic > 0:
            nrmse = rmse / median_organic * 100
        else:
            nrmse = None
        
        # expected total at saturation = the curve's upper limit
        if curve_name == "asymptotic":
            asymptote = parameters[0]
            inflection_passed = True                      # this curve has no inflection point
        else:
            asymptote = parameters[0] + parameters[1]
            inflection_passed = parameters[3] < last_month  # d = month of fastest growth, must be in the past

        fit_rows.append({
            "class": class_name, "curve": curve_name, "fitted": True,
            "rmse": rmse, "asymptote": asymptote,
            "current_organic": current_organic,
            "inflection_passed": inflection_passed,
            "recent_growth": recent_growth,
            "nrmse_percent": nrmse,
        })

fits = pd.DataFrame(fit_rows)
```

Decision and completeness per class

```python
result_rows = []
for class_name in CLASS_COLUMNS:
    class_fits = fits[(fits["class"] == class_name) & (fits["fitted"])]
    events_total = event_volume[class_name]

    # keep curves that pass the checks
    valid = class_fits[class_fits["inflection_passed"]]
    valid = valid[valid["asymptote"] >= valid["current_organic"] * 0.98]   # total can't be below today (2% tolerance)

    recent_growth = None
    if len(class_fits) > 0:
        recent_growth = class_fits["recent_growth"].iloc[0]
    growth_is_low = (not pd.isna(recent_growth)) and recent_growth < RECENT_GROWTH_MAX

    if pd.isna(recent_growth):                       # [lib] True for None or NaN
        reason = "no organic growth left after removing events"
    elif len(valid) == 0:
        reason = "no curve passes the checks"
    elif not growth_is_low:
        reason = "recent growth too high (not saturated)"
    else:
        reason = None
    if reason is not None:
        result_rows.append({"class": class_name, "method": "option B needed",
                            "reason": reason, "recent_growth": recent_growth})
        continue

    # keep only curves that fit about as well as the best one
    best_rmse = valid["rmse"].min()
    valid = valid[valid["rmse"] <= best_rmse * RMSE_TOLERANCE]

    best = valid.sort_values("rmse").iloc[0]
    lowest_total = valid["asymptote"].min()
    highest_total = valid["asymptote"].max()
    spread = (highest_total - lowest_total) / best["asymptote"]

    # add the events back: completeness = (organic today + events) / (organic total + events)
    current_total = best["current_organic"] + events_total
    completeness = current_total / (best["asymptote"] + events_total)
    if completeness > 1:
        completeness = 1.0

    result_rows.append({
        "class": class_name, "method": "saturation (" + best["curve"] + ")",
        "completeness": completeness,
        "completeness_low": current_total / (highest_total + events_total),
        "completeness_high": min(1.0, current_total / (lowest_total + events_total)),
        "curves_agree": spread < STABILITY_MAX, "spread": spread,
        "recent_growth": recent_growth,
    })

completeness_results = pd.DataFrame(result_rows)
print(fits)
completeness_results
```

### Option B (Gröchenig, et al. 2014)

Stablishing parameters

```python
# --- settings according Groechenig et al. (2014) ---
SATURATION_SHARE = 0.03   # <= 3% new in the last year
LOW_GROWTH_MAX = 0.25     #  low growth <= 25%
MEDIUM_GROWTH_MAX = 0.75  #  medium growth 25-75%, high > 75%
MIN_ACTIVE_YEARS = 3      #  saturation needs >= 3 active years
MIN_CONTRIBUTORS = 2      #  growth needs >= 2 active contributors  
```

Calculating mapping stage

```python
window_end = growth["month"].iloc[-1] + pd.DateOffset(months=1) # end of the last month
window_start = window_end - pd.DateOffset(months=RECENT_MONTHS) # 12 months before


stage_rows = []
for class_name, (series_column, class_flag) in CLASS_COLUMNS.items():
    values = growth[series_column].to_numpy(dtype=float)   # total series (events included)
    value_now = values[-1]
    value_year_ago = values[-1 - RECENT_MONTHS]

    # all edits of this class, to count contributors and active years
    class_edits = history[history[class_flag]]

    # contributors in the last 12 months
    recent_edits = class_edits[(class_edits["timestamp"] >= window_start) & (class_edits["timestamp"] < window_end)]
    active_contributors = recent_edits["user"].nunique()

    # active years = calendar years with at least one edit of this class
    active_years = class_edits["timestamp"].dt.year.nunique()

    # share of today's data added in the last 12 months 
    if value_now > 0:
        share_new = (value_now - value_year_ago) / value_now
    else:
        share_new = None

    # decide the stage (saturation is checked first)
    if value_now <= 0:
        stage = "No data"
    elif share_new <= SATURATION_SHARE and active_years >= MIN_ACTIVE_YEARS:
        stage = "Saturation"
    elif active_contributors < MIN_CONTRIBUTORS:
        stage = "Start"
    elif share_new <= LOW_GROWTH_MAX:
        stage = "Low growth"
    elif share_new <= MEDIUM_GROWTH_MAX:
        stage = "Medium growth"
    else:
        stage = "High growth"

    stage_rows.append({
        "class": class_name,
        "share_new_last_year": share_new,
        "active_contributors": active_contributors,
        "active_years": active_years,
        "stage": stage,
    })

option_b = pd.DataFrame(stage_rows)
option_b
```
